```python
import numpy as np
import pandas as pd
df = pd.read_csv("mydataset.csv")
ad_columns = [col for col in df.columns if col.startswith("Ad")]
n_arms = len(ad_columns)
nums_episodes = len(df) 
epsilon = 0.1
q_values = np.zeros(n_arms)
counts = np.zeros(n_arms)
reward_sum = np.zeros(n_arms)
all_rewards = []
action_history = []

for episode in range(nums_episodes):

    if np.random.rand() < epsilon:
        action = np.random.randint(0, n_arms)
    else:
        action = np.argmax(q_values)
    reward = df.loc[episode, ad_columns[action]]

    counts[action] += 1
    reward_sum[action] += reward

    q_values[action] = reward_sum[action] / counts[action]
    all_rewards.append(reward)
    action_history.append(action)


print("Estimated Q-values for each arm:", np.round(q_values, 2))
print("Number of times each arm was selected:", counts.astype(int))
print("Average reward over all episodes:", round(np.mean(all_rewards), 3))
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(10, 6))
plt.hist(
    action_history,
    bins=np.arange(n_arms + 1) - 0.5,
    rwidth=0.8,
    color="skyblue",
    edgecolor="black",
)
plt.title("Epsilon-Greedy Ad Selections (Multi-Armed Bandit)", fontsize=14)
plt.xlabel("Advertisements (Arms)", fontsize=12)
plt.ylabel("Number of Times Selected", fontsize=12)
plt.xticks(range(n_arms), ad_columns, rotation=45)
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.tight_layout()
plt.show()
```
