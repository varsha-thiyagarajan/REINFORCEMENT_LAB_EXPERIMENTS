```python
import numpy as np
from collections import defaultdict
class SimpleEnv:
    def __init__(self): self.reset()
    def reset(self): self.s = 0; return self.s
    def step(self, a):
        self.s = min(self.s + 1, 4) if a == 1 else max(self.s - 1, 0)
        return self.s, (10.0 if self.s == 4 else -1.0), (self.s == 4)

policy = lambda s: np.random.choice([0, 1])
def mc_prediction(env, policy, num_episodes, gamma=0.9):
    V, sums, counts = defaultdict(float), defaultdict(float), defaultdict(float)

    for _ in range(num_episodes):
        ep, s, done = [], env.reset(), False
        while not done:
            a = policy(s)
            nxt, r, done = env.step(a)
            ep.append((s, r))
            s = nxt
        G = 0
        states = [step[0] for step in ep]
        for t in reversed(range(len(ep))):
            s_t, r_t = ep[t]
            G = gamma * G + r_t
            if s_t not in states[:t]:  # First-visit check
                sums[s_t] += G
                counts[s_t] += 1
                V[s_t] = sums[s_t] / counts[s_t]
    return V
env = SimpleEnv()
for s, val in sorted(mc_prediction(env, policy, 2000).items()):
    print(f"State {s}: {val:.2f}")
```
